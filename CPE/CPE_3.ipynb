##Primary Arithmetic

```python
while True:
  a, b = map(int, input().split())
  if a == 0 and b == 0:
    break
  c = cnt = 0
  while a or b:
    s = a% 10 + b % 10 + c
    c = s >= 10
    cnt += c
    a //= 10
    b //= 10
  if cnt == 0:
    print("No carry operation.")
  elif cnt == 1:
    print("1 carry operation.")
  else:
    print(f"{cnt} carry operations.")
```

##Can You Solve It

```python
def f(x,y):
  return(x + y)*(x + y + 1) // 2 + x

t = int(input())
for case in range(1, t+1):
  x1, y1, x2, y2 = map(int, input().split())
  ans = f(x2, y2) - f(x1, y1)
  print(f"Case {case}: {ans}")
```

##Perfection

```python
def sums(n):
    if n == 1:
        return 0

    total = 1
    i = 2

    while i * i <= n:
        if n % i == 0:
            total += i

            if i != n // i:
                total += n // i

        i += 1

    return total


print("PERFECTION OUTPUT")

nums = list(map(int, input().split()))

for n in nums:
    if n == 0:
        break

    s = sums(n)

    if s == n:
        result = "PERFECT"
    elif s < n:
        result = "DEFICIENT"
    else:
        result = "ABUNDANT"

    print(f"{n} {result}")

print("END OF OUTPUT")
```
