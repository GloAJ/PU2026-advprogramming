##Polynomial Showdown

```python
import sys

def solve():
    data = sys.stdin.read().split()
    if not data: return

    for idx in range(0, len(data), 9):
        c = [int(x) for x in data[idx:idx+9]]
        if len(c) < 9: break

        res, first = [], True
        for i in range(9):
            if c[i] == 0: continue
            deg = 8 - i

            if first:
                if c[i] < 0: res.append("-")
            else:
                res.append(" + " if c[i] > 0 else " - ")

            if abs(c[i]) != 1 or deg == 0: res.append(str(abs(c[i])))
            if deg > 0: res.append("x")
            if deg > 1: res.append(f"^{deg}")
            first = False

        print("".join(res) if res else "0")

solve()
```

##Hiaku Review

```python
import sys
def count(s):
    cnt, v = 0, False
    for c in s:
        is_v = c in "aeiouy"
        if is_v and not v: cnt += 1
        v = is_v
    return cnt
for line in sys.stdin:
    line = line.strip()
    if not line or line == "e/o/i": break
    p = line.split('/')
    if count(p[0]) != 5: print("1")
    elif count(p[1]) != 7: print("2")
    elif count(p[2]) != 5: print("3")
    else: print("Y")
```

##Zapping

```python
Import sys
for line in sys.stdin:
    a, b = map(int, line.split())
    if a == -1 and b == -1:
        break
    diff = abs(a - b)
    print(min(diff, 100 - diff))
```

##Word Amalgamation

```python
import sys
def solve():
    a = sys.stdin.read().split()
    if not a: return
    b = a.index("XXXXXX")
    d = sorted(a[:b])
    for s in a[b+1:]:
        if s == "XXXXXX": break
        t = sorted(s)
        res = [w for w in d if sorted(w) == t]
        if res: print("\n".join(res))
        else: print("NOT A VALID WORD")
        print("******")
solve()
```
