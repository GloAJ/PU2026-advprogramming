##Happy Numbers

```
def digits(n):
    return sum(int(d) ** 2 for d in str(n))

def happy(n):
    s = set()
    while n != 1 and n not in s:
        s.add(n)
        n = digits(n)
    return n == 1

t = int(input())
for i in range(1, t + 1):
    n = int(input())

    if happy(n):
        print(f"Case #{i}: {n} is a Happy number.")
    else:
        print(f"Case #{i}: {n} is an Unhappy number.")
```

##Pizza Cutting

```
while True:
    try:
        n = int(input())
        if n < 0:
            break
        print((n*n+n+2)//2)
    except EOFError:
        break
```

##The 3n + 1 problem

```
#include <stdio.h>

int length(long long n) {
    int count = 1;

    while (n != 1) {
        if (n % 2 == 1)
            n = 3 * n + 1;
        else
            n = n / 2;

        count++;
    }

    return count;
}

int main() {
    int i, j;

    while (scanf("%d %d", &i, &j) != EOF) {
        int start = i, end = j;

        if (i > j) {
            int temp = i;
            i = j;
            j = temp;
        }

        int maks = 0;

        for (int k = i; k <= j; k++) {
            int len = length(k);

            if (len > maks)
                maks = len;
        }

        printf("%d %d %d\n", start, end, maks);
    }

    return 0;
}
```

##B2-Sequence

```
case = 1

while True:
    try:
        n = int(input())
        arr = list(map(int, input().split()))

        val = True

        for i in range(n):
            if arr[i] <= 0:
                val = False

            if i > 0 and arr[i] <= arr[i - 1]:
                val = False

        sums = set()

        for i in range(n):
            for j in range(i, n):
                s = arr[i] + arr[j]

                if s in sums:
                    val = False
                else:
                    sums.add(s)

        if val:
            print(f"Case #{case}: It is a B2-Sequence.")
        else:
            print(f"Case #{case}: It is not a B2-Sequence.")

        case += 1

    except:
        break
```
