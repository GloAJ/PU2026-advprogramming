#The 3n + 1 Sequence (Single)

```python
def sequence(n):
    i = []
    while n != 1:
        i.append(str(n))
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
    i.append("1")
    print(" ".join(i))

n = int(input())
sequence(n)
```

#Ant on a 5x5 Chessboard

```python
import sys
import math

def solve():
    intd = sys.stdin.read().split()
    if not intd:
        return
    n = int(intd[0])

    k = math.ceil(math.sqrt(n))
    mid = k * k - k + 1

    if k % 2 != 0:
        if n <= mid:
            x = k
            y = n - (k - 1) * (k - 1)
        else:
            x = k * k - n + 1
            y = k
    else:
        if n <= mid:
            x = n - (k - 1) * (k - 1)
            y = k
        else:
            x = k
            y = k * k - n + 1
    print(f"({x}, {y})")
solve()
```

#Perfect, Deficient, or Abundant

```python
nums = [int(input())]

for n in nums:
    if n == 1:
        s = 0
    else:
        s = 1
        i = 2
        while i * i <= n:
            if n % i == 0:
                s += i
                if i != n // i:
                    s += n // i
            i += 1
    if s == n:
        print("perfect")
    elif s < n:
        print("deficient")
    else:
        print("abundant")
```

#Simple Cipher

```python
import sys
line = sys.stdin.read().splitlines()
n = 0
out = []
while n + 1 < len(line):
    s = line[n].strip()
    t = line[n + 1].strip()
    n += 2
    if not s or not t:
        continue
    ss = [0] * 26
    tt = [0] * 26
    for ch in s:
        ss[ord(ch) - 65] += 1
    for ch in t:
        tt[ord(ch) - 65] += 1
    ss = sorted([x for x in ss if x])
    tt = sorted([x for x in tt if x])
    out.append("YES" if ss == tt else "NO")
sys.stdout.write("\n".join(out) + "\n")
```

#3-Line Haiku Check

```python
l = [input() for _ in range(3)]
t = [5, 7, 5]
v = "aeiouy"
i = 0
while i < 3:
    line = l[i]
    s = 0
    vl = False
    j = 0
    while j < len(line):
        if line[j] in v:
            if not vl:
                s += 1
                vl = True
        else:
            vl = False
        j += 1
    if s != t[i]:
        print(i + 1)
        break
    i += 1
else:
    print("Y")
```

#Parentheses Balance

```python
import sys

def solve():
    intd = sys.stdin.read().splitlines()
    if not intd:
        return

    n = int(intd[0])

    for i in range(1, n + 1):
        s = intd[i].strip() if i < len(intd) else ""

        while "()" in s or "[]" in s:
            s = s.replace("()", "").replace("[]", "")

        if s == "":
            print("Yes")
        else:
            print("No")
solve()
```

#Cool Word

```python
import sys

def solve():
    data = sys.stdin.read().split()
    if not data: return
    it = iter(data)
    case = 1

    for n_str in it:
        cool = 0
        for _ in range(int(n_str)):
            w = next(it)
            f = [w.count(c) for c in set(w)]
            if len(f) >= 2 and len(f) == len(set(f)):
                cool += 1
        print(f"Case {case}: {cool}")
        case += 1

solve()
```
