#Reverse Iteration

```python
class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        for i in range(len(digits) - 1, -1, -1):
            if digits[i] < 9:
                digits[i] += 1
                return digits
            digits[i] = 0
        return [1] + digits
```

#Type Conversion

```python
class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        n = "".join(map(str, digits))
        new = str(int(n) + 1)
        return [int(char) for char in new]
```

#While Loop

```python
class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        i = len(digits) - 1

        while i >= 0:
            if digits[i] == 9:
                digits[i] = 0
                i -= 1
            else:
                digits[i] += 1
                return digits
        digits.insert(0, 1)
        return digits
```
