Basic wall maze

```
import collections

def solve():
    while True:
        try:
            line = input().split()
            if not line or line == ['0', '0']: break
            s, e = tuple(map(int, line)), tuple(map(int, input().split()))

            walls = set()
            for _ in range(3):
                x1, y1, x2, y2 = map(int, input().split())
                if x1 == x2:
                    for y in range(min(y1, y2) + 1, max(y1, y2) + 1):
                        walls.add(((x1, y), (x1+1, y)))
                        walls.add(((x1+1, y), (x1, y)))
                else:
                    for x in range(min(x1, x2) + 1, max(x1, x2) + 1):
                        walls.add(((x, y1), (x, y1+1)))
                        walls.add(((x, y1+1), (x, y1)))

            q = collections.deque([(s, "")])
            visited = {s}
            while q:
                (cx, cy), path = q.popleft()
                if (cx, cy) == e:
                    print(path)
                    break

                for dx, dy, d in [(0,-1,'N'), (1,0,'E'), (0,1,'S'), (-1,0,'W')]:
                    nx, ny = cx + dx, cy + dy
                    if 1 <= nx <= 6 and 1 <= ny <= 6 and ((cx,cy), (nx,ny)) not in walls and (nx,ny) not in visited:
                        visited.add((nx, ny))
                        q.append(((nx, ny), path + d))
        except EOFError: break

solve()
```

Ancient Cipher

```
import sys

lines = sys.stdin.read().splitlines()
i = 0
out = []
while i + 1 < len(lines):
    s = lines[i].strip()
    t = lines[i+1].strip()
    i += 2
    if not s or not t:
        continue
    fs = [0] * 26
    ft = [0] * 26
    for ch in s:
        fs[ord(ch) - 65] += 1
    for ch in t:
        ft[ord(ch) - 65] += 1
    fs = sorted([x for x in fs if x])
    ft = sorted([x for x in ft if x])
    out.append("YES" if fs == ft else "NO")
sys.stdout.write("\n".join(out))
```

Checkers

```
import sys

i = sys.stdin.read().split()
if i:
    T = int(i[0])
    idx = 1
    MOD = 1000007

    for n in range(1, T + 1):
        N = int(i[idx])
        grid = i[idx + 1 : idx + 1 + N][::-1]
        idx += 1 + N

        sy, sx = next((y, x) for y in range(N) for x in range(N) if grid[y][x] == 'W')
        dp = [[0] * N for _ in range(N)]
        dp[sy][sx] = 1

        for y in range(N - 1):
            for x in range(N):
                if not dp[y][x]: continue

                for d in (-1, 1):
                    nx, ny = x + d, y + 1
                    if 0 <= nx < N:
                        if grid[ny][nx] == '.':
                            dp[ny][nx] = (dp[ny][nx] + dp[y][x]) % MOD
                        elif grid[ny][nx] == 'B':
                            jx, jy = x + 2 * d, y + 2
                            if 0 <= jx < N and jy < N and grid[jy][jx] == '.':
                                dp[jy][jx] = (dp[jy][jx] + dp[y][x]) % MOD

        print(f"Case {n}: {sum(dp[N - 1]) % MOD}")
```

Finding Haplotypes

```
import sys

def make_pairs(g):
    pos = [i for i, c in enumerate(g) if c == '2']
    pairs = []

    for mask in range(1 << len(pos)):
        a = list(g)
        b = list(g)

        for i in range(len(g)):
            if g[i] == '0':
                a[i] = b[i] = '0'
            elif g[i] == '1':
                a[i] = b[i] = '1'

        for j, p in enumerate(pos):
            if (mask >> j) & 1:
                a[p] = '0'
                b[p] = '1'
            else:
                a[p] = '1'
                b[p] = '0'

        pairs.append(("".join(a), "".join(b)))

    return pairs


data = sys.stdin.read().strip().split()
idx = 0

while idx < len(data):
    N = int(data[idx])
    L = int(data[idx + 1])
    idx += 2

    G = data[idx:idx + N]
    idx += N

    options = [make_pairs(g) for g in G]

    ans = 10**9

    def dfs(i, used):
        global ans

        if len(used) >= ans:
            return

        if i == N:
            ans = min(ans, len(used))
            return

        for a, b in options[i]:
            new_used = used | {a, b}
            dfs(i + 1, new_used)

    dfs(0, set())

    print(ans)
```
